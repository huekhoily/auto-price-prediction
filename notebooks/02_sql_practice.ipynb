# SQL Analysis

This notebook uses SQL queries to explore the cleaned used-car database stored in SQLite.

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/cars.db")
```

```python
query1 = """
SELECT 
	brand,
	COUNT(*) as n_cars
FROM cars
GROUP BY brand
ORDER BY n_cars DESC
LIMIT 10;
"""

pd.read_sql_query(query1, conn)
```

Ford has the highest number of listings in the dataset, with 386 vehicles, followed by BMW with 375 vehicles.

```python
query2 = """
SELECT
	fuel_type,
	COUNT(*) AS n_cars,
    ROUND(AVG(price), 2) AS average_price
FROM cars
GROUP BY fuel_type
HAVING COUNT(*) >= 30
ORDER BY average_price DESC;
"""

pd.read_sql_query(query2, conn)
```

Among fuel types with at least 30 listings, Hybrid had the highest average listed price at about 51426, while E85 Flex Fuel had the lowest average price at approximately 22744.

```python
query3 = """
SELECT 
	accident,
	COUNT(*) AS n_cars,
	ROUND(AVG(price), 2) AS average_price
FROM cars
GROUP BY accident
ORDER BY average_price DESC;
"""

pd.read_sql_query(query3, conn)
```

Vehicles with reported accident or damage history had a lower average listed price than vehicles with no reported accidents.

```python
query4 = """
SELECT
	transmission_group,
	COUNT(*) AS n_cars,
	ROUND(AVG(price), 2) AS average_price
FROM cars
GROUP BY transmission_group
HAVING COUNT(*) >= 50
ORDER BY average_price DESC;
"""

pd.read_sql_query(query4, conn)
```

Dual Shift has highest average listed price at approximately 54711. However, automatic has by far the most listings, with 3110 vehicles, and has the second-highest average listed price at 43895.

```python
query5 = """
SELECT
    brand,
    COUNT(*) AS n_cars,
    ROUND(AVG(price), 2) AS average_price
FROM cars
WHERE model_year >= 2020
GROUP BY brand
HAVING COUNT(*) >= 20
ORDER BY average_price DESC;
"""

pd.read_sql_query(query5, conn)
```

Porsche has the highest average listed price at approximately 124579, followed by Mercedes-Benz at 95154.

```python
query6 = """
SELECT brand, model, model_year, mileage, price
FROM cars
WHERE model_year >= 2020
  AND mileage >= 100000
ORDER BY mileage DESC
LIMIT 15;
"""

pd.read_sql_query(query6, conn)
```

Several vehicles from 2020 or later have mileage above 100000 miles. 

```python
query7 = """
SELECT brand, model, model_year, mileage, price
FROM cars
WHERE model_year >= 2020
  AND mileage < 30000
  AND price BETWEEN 20000 AND 40000
ORDER BY price ASC
LIMIT 15;
"""

pd.read_sql_query(query7, conn)
```

These listings may represent relatvively affordable options among newer vehicles with lower mileage.
